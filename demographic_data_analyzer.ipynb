```python
import pandas as pd
```

```python
columns = ['age', 'workclass', 'fnlwgt', 'education', 'education-num', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income']

df = pd.read_csv('adult.data', header=None, names=columns, skipinitialspace=True)

print(len(df))
print(df.head())
```

```python
print(df.info())
```

```python
men = df[df['sex'] == 'Male']
avg_mean = men['age'].mean()
print("avarage men age", avg_mean)
```

```python
edu = df['education'].value_counts()
edu_percentage = df['education'].value_counts(normalize=True) * 100
# print(edu)

print("Percentage of bchelor degree =", edu_percentage)
print("Percentage of bchelor degree =", edu_percentage['Bachelors'])
```

```python
advance = df[df["education"].isin(['Bachelors', 'Masters', 'Doctorate'])]
percentage_income = round(
    (advance['income'] == '>50K').sum() / len(advance) *  100, 4
)

print("percentage of advance degree more than 50K =", percentage_income)
```

```python
common = df[~df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]
income_common = round(
    (common['income'] == '>50K').sum() / len(common) * 100, 4
)

print("percentage of common degree more than 50K =", income_common)
```

```python
minimum_hours = df['hours-per-week'].min()
print("Minimum hour per-week =", minimum_hours)
```

```python
minimum_income = df[df['hours-per-week'].isin([1])]

result = round(
    (minimum_income['income'] == ">50K").sum() / len(minimum_income) * 100, 4
)

print("percentage people had over 50K with minimum hour work =", result)
```

```python
country_income = df.groupby("native-country").apply(lambda x: (x["income"] == ">50K").sum() / len(x) * 100)
top_country = country_income.idxmax()
top_percentage = round(country_income.max(), 4)

print("Top country =", top_country)
print("Top percentage =", top_percentage)
```

```python
india_top_job = df[(df['native-country'] == 'India') & (df['income'] == '>50K')]
top_job = india_top_job['occupation'].value_counts().idxmax()
print("most popular occupation in india with >50K income is", top_job)
```
